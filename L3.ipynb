## Instrukcje warunkowe

Z warunkami spotykamy się w każdej dziedzinie życia. Na pytanie "Czy pada deszcz?" odpowiedzią może być "Tak" lub "Nie". W zależności od odpowiedzi wiemy jak się ubrać lub czy wziąć parasol. Podobnie w programowaniu, instrukcje warunkowe pozwalają na wykonanie różnych bloków kodu w zależności od spełnienia określonych warunków.

W Pythonie kluczowe słowa służące do tworzenia instrukcji warunkowych to `if` i `else`. Po słowie kluczowym dodajemy dwukuropek, a instrukcje które mają się wykonać po danym warunku umieszczamy w bloku kodu z wcięciem wykonywanym za pomocą tabulacji.

![Space](space.jpg)

Składnia instrukcji:
```Python
if warunek:
    #kod, gdy warunek jest spełniony
else:
    #kod, na wszystko inne
```

Wypróbuj przykład poniżej dla pogody "pada" i innej

````python
pogoda = "pada"
if pogoda == "pada":
    print("Weź parasol byczku")
else:
    print("Nie bierz parasola")
````

Instrukcja `if` sprawdza wynik warunku (warunek zwraca wartość _True_ lub _False_). `If` może sprawdzić czy podana wartość jest równa innej wartości, lub czy jest większa/mniejsza tak jak w poniższej tabelce:
| Operator | Nazwa po polsku |
| --- | --- |
| `a == b` | Równe |
| `a != b` | Różne |
| `a < b` | Mniejsze niż |
| `a <= b` | Mniejsze lub równe |
| `a > b` | Większe niż |
| `a >= b` | Większe lub równe |

## Losowanie liczb

Zacznijmy od pokazania jak działa losowanie liczb.
