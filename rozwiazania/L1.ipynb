## Instrukcja print
Print to instrukcja w języku Python, która służy do wypisywania tekstu na ekranie. Za pomocą tej instrukcji możemy wypisać jakiś tekst lub wyrażenie na ekran komputera. Komunikat który chcemy wypisać umieszczamy w środku okrągłego nawiasu tak jak w przykładzie powyżej.

### Ćwiczenie 1
Jaka jest różnica między instrukcjami print poniżej?

```python
print("Cześć!")
print("4+2")
print(4+2)
```

1. `print("Cześć!")` – wypisuje na ekran tekst Cześć!.
2. `print("4+2")` – wypisuje na ekran tekst 4+2, ponieważ działanie znajduje się w cudzysłowie i Python traktuje je jako zwykły tekst.
3. `print(4+2)` – wykonuje działanie matematyczne 4 + 2 i wypisuje jego wynik, czyli 6.
Tekst umieszczony w cudzysłowie jest wypisywany dokładnie tak, jak został zapisany, natomiast wyrażenie bez cudzysłowu jest interpretowane i obliczane.
