## Zmienne

### Czym jest zmienna?

Zmienna to nazwana część pamięci komputera, w której przechowywana jest jakaś wartość.

Zmienną można sobie wyobrazić jako pudełko, w którym możemy umieścić różne rzeczy i nakleić na nie etykietę z nazwą. Zamiast za każdym razem pisać skomplikowaną wartość, tworzysz zmienną (pudełko z etykietą) i od teraz używasz jej prostej nazwy.

>Obowiązują pewne ograniczenia dotyczące znaków, które mogą być używane w nazwach zmiennych Pythona. Dozwolone są tylko litery, cyfry i podkreślenia. Ponadto nazwy nie mogą zaczynać się od liczb. Nieprzestrzeganie tych zasad powoduje błędy.


### Typy danych
W Pythonie wszystko jest jakimś typem danych, tak jak klocki Lego - różne kształty służą do różnych celów.
W tabelce znajdują się cztery podstawowe typy, których będziemy używać:

| Typ            | Co to jest?                | Przykład                     | Przykład z życia                                      |
|----------------|----------------------------|------------------------------|-------------------------------------------------------|
| int (integer)  | Liczba całkowita.          | `2`, `100`, `-50`            | Twój wiek, liczba uczniów w klasie, numer buta.       |
| float          | Liczby zmiennoprzecinkowe. | `19.99`, `4.5`, `-10.2`      | Cena produktu, średnia ocen, temperatura na zewnątrz. |
| str (string)   | Dowolny tekst.             | `"Ania"`, `'Poprawny login'` | Twoje imię, tytuł filmu, treść SMS-a, hasło.          |
| bool (boolean) | Tylko prawda lub fałsz.    | True, False                  | Tak, nie.                                             |
> Tekst (string) zawsze umieszczamy w cudzysłowiu, lub w apostrofach.

## Wprowadzanie danych
Gdy korzystamy z jakiegoś programu albo aplikacji może wystąpić sytuacja, w której użytkownik musi podać jakieś dane np. login i hasło, albo swoją datę urodzenia. W Pythonie do takich operacji służy instrukcja `input()`. Wypróbuj przykłady z poniższego bloku:

```python
login = input("Podaj login: ")
print("Twój login to: " , login)

haslo = input("Podaj hasło: ")
print("Twoje hasło to: ", haslo)

rok_urodzenia = int(input("Podaj rok urodzenia: "))
print("Twój rok urodzenia to: ", rok_urodzenia)
```

### Ćwiczenie 4
Napisz program, który zapyta użytkownika o dwie liczby `x`, `y`, a następnie zapisze po kolei do zmiennej `wynik` wynik działania a w instrukcji `print()` wypisze podane liczby oraz wynik działania.
- Dodawanie (+)
- Odejmowanie (-)
- Mnożenie (*)
- Dzielenie (/)

```python
# Rozwiązanie - ćw. 4
x = int(input("Podaj x: "))
y = int(input("Podaj y: "))
wynik = x + y
print(x, "+", y, "=", wynik)

wynik = x + y
print(x, "+", y, "=", wynik)

wynik = x * y
print(x, "*", y, "=", wynik)

wynik = x - y
print(x, "-", y, "=", wynik)

wynik = x / y
print(x, "/", y, "=", wynik)
```
